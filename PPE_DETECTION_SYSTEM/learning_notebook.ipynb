```python
import cv2

cap = cv2.VideoCapture(0)

# Check if camera opened
if not cap.isOpened():
    print("Camera could not be opened or permission was denied.")
    cap.release()
    cv2.destroyAllWindows()
    exit()

while True:
    ret, frame = cap.read()

    # If frame wasn't captured, exit gracefully
    if not ret:
        print("Failed to capture frame. Camera may have been closed or permission denied.")
        break

    cv2.imshow("Camera", frame)

    key = cv2.waitKey(1) & 0xFF

    # ESC or window closed
    if key == 27 or cv2.getWindowProperty("Camera", cv2.WND_PROP_VISIBLE) < 1:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

img = cv2.imread("office.png")

# Resize
resize = cv2.resize(img, (600, 400))

# Crop
crop = resize[50:300, 100:400]

# Gray
gray = cv2.cvtColor(resize, cv2.COLOR_BGR2GRAY)

# Blur
blur = cv2.GaussianBlur(gray, (7, 7), 0)

# Threshold
_, thresh = cv2.threshold(blur, 127, 255, cv2.THRESH_BINARY)

# Edge Detection
edges = cv2.Canny(blur, 100, 200)

cv2.imshow("Original", resize)
cv2.imshow("Crop", crop)
cv2.imshow("Gray", gray)
cv2.imshow("Blur", blur)
cv2.imshow("Threshold", thresh)
cv2.imshow("Edges", edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Cannot open camera")
    exit()

cv2.namedWindow("Camera")
cv2.namedWindow("ROI")

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Draw ROI
    cv2.rectangle(frame, (180, 80), (420, 380), (0, 255, 0), 2)

    # Extract ROI
    roi = frame[80:380, 180:420]

    cv2.imshow("Camera", frame)
    cv2.imshow("ROI", roi)

    key = cv2.waitKey(1) & 0xFF

    # Exit on ESC
    if key == 27:
        break

    # Exit if either window is closed
    if (cv2.getWindowProperty("Camera", cv2.WND_PROP_VISIBLE) < 1 or
        cv2.getWindowProperty("ROI", cv2.WND_PROP_VISIBLE) < 1):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# Mini Project 
import cv2
import time

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Cannot open camera.")
    exit()

prev_time = time.time()

while True:

    ret, frame = cap.read()

    if not ret:
        print("Failed to capture frame.")
        break

    # -----------------------------
    # Calculate FPS
    # -----------------------------
    current_time = time.time()
    fps = 1 / (current_time - prev_time)
    prev_time = current_time

    # -----------------------------
    # Draw Rectangle
    # -----------------------------
    cv2.rectangle(frame,
                  (150, 100),
                  (350, 400),
                  (0, 255, 0),
                  2)

    # -----------------------------
    # Show FPS
    # -----------------------------
    cv2.putText(frame,
                f"FPS: {int(fps)}",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0, 255, 255),
                2)

    # Instructions
    cv2.putText(frame,
                "Press S = Save | ESC = Exit",
                (20, 470),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (255, 255, 255),
                2)

    # Show Webcam
    cv2.imshow("Mini Project 1", frame)

    key = cv2.waitKey(1) & 0xFF

    # -----------------------------
    # Save Screenshot
    # -----------------------------
    if key == ord('s'):
        filename = f"screenshot_{int(time.time())}.jpg"
        cv2.imwrite(filename, frame)
        print(f"Saved: {filename}")

    # -----------------------------
    # Exit on ESC
    # -----------------------------
    elif key == 27:
        break

    # Exit if window is closed
    if cv2.getWindowProperty("Mini Project 1", cv2.WND_PROP_VISIBLE) < 1:
        break

# Cleanup
cap.release()
cv2.destroyAllWindows()
```

# Topic: YOLO Webcam Detection

```python
from ultralytics import YOLO
import cv2

# Load YOLO model
model = YOLO("yolo11n.pt")

# Open webcam
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # YOLO prediction
    results = model(frame)

    # Draw detections
    annotated_frame = results[0].plot()

    # Show output
    cv2.imshow(
        "YOLO Live Detection",
        annotated_frame
    )

    # ESC to exit
    if cv2.waitKey(1) & 0xFF == 27:
        break


cap.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

model = YOLO("yolo11n.pt")

img = cv2.imread("office.png")

results = model(img, classes=[0])

boxes = results[0].boxes


employee_id = 1

for box in boxes:

    # Get coordinates
    x1, y1, x2, y2 = box.xyxy[0]

    x1 = int(x1)
    y1 = int(y1)
    x2 = int(x2)
    y2 = int(y2)

    # Draw bounding box
    cv2.rectangle(
        img,
        (x1,y1),
        (x2,y2),
        (0,255,0),
        2
    )

    # Draw ID
    cv2.putText(
        img,
        f"Employee {employee_id}",
        (x1,y1-10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0,255,255),
        2
    )

    employee_id += 1


cv2.imshow("Employees", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

# Load YOLO model
model = YOLO("yolo11m.pt")


# Open webcam
cap = cv2.VideoCapture(0)


while True:

    ret, frame = cap.read()

    if not ret:
        break


    # YOLO + ByteTrack
    results = model.track(
        frame,
        classes=[0],      # only person
        tracker="bytetrack.yaml",
        persist=True
    )


    # Draw results
    annotated_frame = results[0].plot()


    cv2.imshow(
        "Employee Tracking",
        annotated_frame
    )


    # ESC exit
    if cv2.waitKey(1) & 0xFF == 27:
        break


cap.release()
cv2.destroyAllWindows()
```
